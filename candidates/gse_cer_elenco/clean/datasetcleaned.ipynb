```python
import pandas as pd

# 1. Normalizzazione nomi colonne
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')

# 2. Pulizia testo
string_cols = ['tipologia_di_configurazione', 'denominazione_comunità', 'comune', 'provincia', 'regione', 'ragione_sociale_gdr']
for col in string_cols:
    df[col] = df[col].astype(str).str.strip()

# 3. Conversione Latitudine, Longitudine e Potenza in float (gestendo eventuali virgole)
float_cols = ['latitudine', 'longitudine', 'potenza_totale_kw']
for col in float_cols:
    df[col] = df[col].astype(str).str.replace(',', '.').astype('float64')

# 4. Conversione Data
df['data_di_aggiornamento'] = pd.to_datetime(df['data_di_aggiornamento'],dayfirst=True)
```

```python
df.info()
```

```python
df.head()
```

```python
# 1. Applicare strip() a tutte le colonne di testo
string_columns = df.select_dtypes(include=['object', 'string']).columns
for col in string_columns:
    df[col] = df[col].astype(str).str.strip()

# 2. Sigle province in maiuscolo
if 'provincia' in df.columns:
    df['provincia'] = df['provincia'].str.upper()
```

```python
print(f"Righe duplicate: {df.duplicated().sum()}")
```

```python
# Mostra sia l'originale che il duplicato ordinati affiancati
duplicati = df[df.duplicated(keep=False)].sort_values(by=list(df.columns))
duplicati
```

```python
# Rimuove la riga duplicata
df = df.drop_duplicates(keep='first').reset_index(drop=True)

# Verifica rapida del nuovo totale righe (deve passare da 904 a 903)
print(f"Nuovo numero di righe: {len(df)}")
```

```python
df.to_parquet('elenco_comunita_energetiche_rinnovabili.parquet', index=False)
```
