```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots


#  Aggregazioni per Regione
reg_df = df.groupby('regione').agg(
    tot_potenza_kw=('potenza_totale_kw', 'sum'),
    tot_impianti=('numero_impianti', 'sum')
).reset_index()

reg_kw = reg_df.sort_values(by='tot_potenza_kw', ascending=False)
reg_imp = reg_df.sort_values(by='tot_impianti', ascending=False)

#  Aggregazioni per Comune (Top 15)
com_df = df.groupby(['comune', 'provincia', 'regione']).agg(
    tot_potenza_kw=('potenza_totale_kw', 'sum'),
    tot_impianti=('numero_impianti', 'sum')
).reset_index()

com_kw = com_df.sort_values(by='tot_potenza_kw', ascending=False).head(15)
com_imp = com_df.sort_values(by='tot_impianti', ascending=False).head(15)

#  Creazione della griglia 2x2 per i 4 grafici
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        "Potenza Totale (kW) per Regione",
        "Numero Impianti per Regione",
        "Top 15 Comuni per Potenza (kW)",
        "Top 15 Comuni per N. Impianti"
    ),
    vertical_spacing=0.22,
    horizontal_spacing=0.08
)

# --- Grafico 1: Regioni - Potenza (kW) --- [Riga 1, Colonna 1]
fig.add_trace(
    go.Bar(
        x=reg_kw['regione'], 
        y=reg_kw['tot_potenza_kw'],
        marker_color='#1f77b4',
        name='Potenza kW (Regioni)'
    ),
    row=1, col=1
)

# --- Grafico 2: Regioni - N. Impianti --- [Riga 1, Colonna 2]
fig.add_trace(
    go.Bar(
        x=reg_imp['regione'], 
        y=reg_imp['tot_impianti'],
        marker_color='#2ca02c',
        name='N. Impianti (Regioni)'
    ),
    row=1, col=2
)

# --- Grafico 3: Comuni - Potenza (kW) --- [Riga 2, Colonna 1]
fig.add_trace(
    go.Bar(
        x=com_kw['comune'], 
        y=com_kw['tot_potenza_kw'],
        marker_color='#17becf',
        name='Potenza kW (Comuni)'
    ),
    row=2, col=1
)

# --- Grafico 4: Comuni - N. Impianti --- [Riga 2, Colonna 2]
fig.add_trace(
    go.Bar(
        x=com_imp['comune'], 
        y=com_imp['tot_impianti'],
        marker_color='#ff7f0e',
        name='N. Impianti (Comuni)'
    ),
    row=2, col=2
)

# 5. Configurazione del layout e stili
fig.update_layout(
    title_text="Quadro Generale CER: Confronto per Regione e Comune (Potenza vs Impianti)",
    title_x=0.5,
    title_font_size=18,
    showlegend=False,
    template="plotly_white",
    height=800,
    margin=dict(l=50, r=50, t=80, b=100)
)

# Inclinazione delle etichette degli assi X per leggibilità
fig.update_xaxes(tickangle=-45)

# Visualizzazione e salvataggio
fig.show()
fig.write_html("quadro_generale_cer_2x2.html")
```

```python
import pandas as pd
import plotly.graph_objects as go

# 1. Ordinamento dei dati per potenza media ad impianto
df_sorted = reg_metriche_complete.sort_values(by='kw_per_impianto', ascending=False)

# 2. Creazione della figura interattiva
fig = go.Figure()

# Barra 1: Potenza Media ad Impianto (kW)
fig.add_trace(go.Bar(
    x=df_sorted['regione'],
    y=df_sorted['kw_per_impianto'],
    name='Potenza Media (kW/Impianto)',
    marker_color='#1f77b4',
    text=df_sorted['kw_per_impianto'],
    textposition='auto'
))

# Barra 2: Utenze Medie per Impianto
fig.add_trace(go.Bar(
    x=df_sorted['regione'],
    y=df_sorted['utenze_per_impianto'],
    name='Utenze Medie / Impianto',
    marker_color='#ff7f0e',
    text=df_sorted['utenze_per_impianto'],
    textposition='auto'
))

# 3. Layout e personalizzazione
fig.update_layout(
    title='Analisi Dettagliata per Regione: Potenza Media vs Utenze per Impianto',
    title_x=0.5,
    barmode='group',  # Barre affiancate
    xaxis_title='Regione',
    yaxis_title='Valore Medio',
    template='plotly_white',
    height=600,
    legend=dict(x=0.7, y=0.95)
)

fig.update_xaxes(tickangle=-45)

# Visualizza e salva il file interattivo
fig.show()
fig.write_html("potenza_e_utenze_medie_regionali.html")
```

```python

#  Definizione Fasce di Potenza
def fascia_potenza(kw):
    if kw < 50:
        return '1. Micro (<50 kW)'
    elif kw <= 200:
        return '2. Piccola (50-200 kW)'
    elif kw <= 500:
        return '3. Media (200-500 kW)'
    else:
        return '4. Grande (>500 kW)'

df['fascia_potenza'] = df['potenza_totale_kw'].apply(fascia_potenza)

# Aggregazione per fascia
cluster_potenza = df.groupby('fascia_potenza').agg(
    numero_cer=('denominazione_comunità', 'count'),
    tot_potenza_kw=('potenza_totale_kw', 'sum'),
    tot_utenze=('numero_utenze', 'sum')
).reset_index()

cluster_potenza['quota_cer_%'] = (cluster_potenza['numero_cer'] / len(df) * 100).round(1)

print("=== SEGMENTAZIONE CER PER FASCE DI POTENZA ===")
print(cluster_potenza.to_string(index=False))
```

```python
import plotly.express as px
#Aggregazione per Regione e Fascia
df_grouped = df.groupby(['regione', 'fascia_potenza']).size().reset_index(name='numero_cer')

#  Grafico a Barre Impilate
fig = px.bar(
    df_grouped, 
    x='regione', 
    y='numero_cer', 
    color='fascia_potenza',
    title='Distribuzione delle CER per Regione e Fascia di Potenza',
    labels={'regione': 'Regione', 'numero_cer': 'Numero di CER', 'fascia_potenza': 'Fascia di Potenza'},
    color_discrete_map={
        '1. Micro (<50 kW)': '#2ca02c',       # Verde
        '2. Piccola (50-200 kW)': '#1f77b4',   # Blu
        '3. Media (200-500 kW)': '#ff7f0e',   # Arancione
        '4. Grande (>500 kW)': '#d62728'       # Rosso
    }
)

fig.update_layout(
    template='plotly_white',
    height=600,
    xaxis={'categoryorder': 'total descending'}, # Ordina le regioni per numero totale di CER
    legend_title_text='Fascia Potenza'
)

fig.update_xaxes(tickangle=-45)

fig.show()
fig.write_html("fasce_potenza_per_regione.html")
```

```python
# Classificazione per dimensione della comunità
def fascia_partecipazione(utenze):
    if utenze <= 5:
        return '1. Ristretta (1-5 utenze)'
    elif utenze <= 20:
        return '2. Media (6-20 utenze)'
    elif utenze <= 50:
        return '3. Ampia (21-50 utenze)'
    else:
        return '4. Sociale/Allargata (>50 utenze)'

df['fascia_partecipazione'] = df['numero_utenze'].apply(fascia_partecipazione)

# Aggregazione
dim_summary = df.groupby('fascia_partecipazione').agg(
    tot_cer=('denominazione_comunità', 'count'),
    tot_utenze=('numero_utenze', 'sum'),
    tot_potenza_kw=('potenza_totale_kw', 'sum')
).reset_index()

dim_summary['quota_cer_%'] = (dim_summary['tot_cer'] / len(df) * 100).round(1)

print("=== PARTECIPAZIONE SOCIALE NELLE CER ===")
print(dim_summary.to_string(index=False))
```
